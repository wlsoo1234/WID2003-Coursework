## What an eye tracker actually does

An eye tracker is a small device (often just below the screen) that uses an infrared camera to follow the reflection of light in your eyes, many times per second. In this study it recorded roughly **120 times every second** — over a hundred data points for every single second a student looked at the screen.

From that raw stream, a few useful ideas emerge. Here they are, explained with everyday analogies:

| Term you'll see | What it actually means |
|---|---|
| **Fixation** | A brief pause where your eyes stop and focus on one spot — like pausing to actually *look* at a word instead of skimming past it. Most of what your eyes do is jump between fixations. |
| **Dwell time** (or **dwell ratio**) | How much of your total looking-time was spent on one particular spot — like asking "of the 10 seconds you spent searching, how many were spent staring at the actual hiding spot?" |
| **AOI** (Area of Interest) | The specific "zone" on the picture that researchers cared about — in this study, that's the region where the hidden object actually was, plus a handful of *decoy* regions nearby that looked plausible but were wrong. |
| **Scanpath** | The trail your eyes left behind — literally the path connecting one fixation to the next, like footprints showing where someone walked while searching a room. |

## How we knew who got the puzzle right

While the eye tracker recorded gaze, students also **clicked on the screen** once they thought they'd found the answer. That click was automatically checked against the known correct location — no manual grading needed. Across the 130 students and 13 puzzles, that gave us **1,690 individual attempts**, each one labelled correct or incorrect.
