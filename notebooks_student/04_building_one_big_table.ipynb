## Background, Explained Simply

### Reshaping the table

```python
# Run this cell to see the real reshape: 1,690 rows (one per student per puzzle)
# become 130 rows (one per student), by spreading each puzzle's measurements
# out into their own columns.
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(11, 5))

ax = axes[0]
ax.bar(['Before\n(one row per\nstudent per puzzle)', 'After\n(one row per\nstudent)'],
       [1690, 130], color=['#3498db', '#2ecc71'])
ax.set_ylabel('Number of rows in the table')
ax.set_title('Reshaping the data')
for i, v in enumerate([1690, 130]):
    ax.text(i, v + 20, str(v), ha='center', fontweight='bold')

ax = axes[1]
ax.bar(['Right after\nreshaping', 'After removing\nmostly-empty columns'],
       [446, 416], color=['#3498db', '#2ecc71'])
ax.set_ylabel('Number of measurement columns')
ax.set_title('Removing columns with too much\nmissing data (30 columns dropped)')
for i, v in enumerate([446, 416]):
    ax.text(i, v + 3, str(v), ha='center', fontweight='bold')
ax.set_ylim(0, 470)

plt.tight_layout()
plt.show()
```

Same information, completely reorganised: instead of 13 separate rows for one student (one per puzzle), that student now gets exactly **one** row, with each puzzle's measurements spread out into their own named columns (e.g. "how long they dwelled on the right spot, *specifically on the Crown puzzle*"). This is what unlocks comparing whole students against each other, rather than individual puzzle attempts.

### Dealing with missing measurements

Not every student has a usable value for every single measurement — for instance, a measurement based on mouse-click timing simply doesn't exist for a student who never clicked at all on one puzzle. Two different fixes were used depending on how bad the gap was:

- If a column was missing for **more than half** of all students, the whole column was **removed** — with that much missing, the column carries too little real information to be worth keeping (30 columns were removed this way).
- For every column that was kept, any remaining gaps were filled in with that column's **median** (its exact middle value) — a deliberately unglamorous, conservative choice. It doesn't invent an unrealistic answer, and it doesn't wildly swing the analysis toward one student's odd value.

### Taming extreme values, without deleting anyone

Occasionally, one student would have a wildly extreme measurement (say, a scanpath ten times longer than nearly everyone else's) — often just from an unusual noisy moment during recording. Rather than deleting that student's whole row over one strange number, extreme values were **capped**: pulled in to a reasonable maximum/minimum, using a standard statistical rule (values further than about one and a half times the "typical spread" from the middle 50% of the data get pulled back to that boundary). The student stays in the study; only the single extreme number is tempered.

### Setting aside a "final exam" group

Before doing anything else with the data, **26 of the 130 students (20%) were set aside** and not touched again until the very last, final check in Stage 6. This mirrors a study habit worth knowing about broadly: if you let yourself peek at your test questions while you're still studying, you'll do suspiciously well on the test — but for the wrong reason. Keeping this "final exam" group **completely untouched** during every decision made along the way is what makes the eventual final check honest.

## What Actually Happened, By the Numbers

- The reshaping step alone (before dropping anything) produced **130 rows × 447 columns**.
- After merging in each student's labels: **130 rows × 465 columns**.
- Of the 446 actual measurement columns, **30 were dropped** for missing more than half their values, leaving **416** — every one of which had any remaining gaps filled with its median.
- The 130 students were split, keeping each group's High/Low balance proportional: **104 for building and choosing the model, 26 held back untouched** as the final check.
- After all cleaning, capping, and standardising each measurement onto a common scale, the final table was **130 rows × 852 columns** (416 real measurements, plus an unscaled copy of each, plus labels and bookkeeping columns).
