## Overview

To compare "strong" and "struggling" problem-solvers, we first need an actual, defensible way to split 130 real students into two groups. This stage explains exactly how that was done — and is honest about the fact that *any* such method involves a judgment call.

## Background, Explained Simply

### How we knew if an answer was right

Students didn't fill in a written answer — they **clicked directly on the picture** where they thought the hidden object was. That click was automatically compared against the one correct "zone" recorded for that puzzle. No manual grading, no ambiguity: either the click landed in the correct zone or it didn't.

Each student's **accuracy** = number of puzzles they got right, out of all 13.

### The median split, explained without formulas

Line up all 130 students from lowest accuracy to highest. Find the middle point (the **median**) — half the students scored at or above it, half scored below. Everyone at or above that middle point is labelled **High-performing**; everyone below is labelled **Low-performing**.

This method has a real strength: it automatically produces two reasonably balanced groups, no matter what the overall score distribution looks like. But it also has a real weakness worth sitting with: **a student scoring exactly at the midpoint and a student scoring one puzzle below it get sorted into different groups**, even though their actual performance was nearly identical. The label is a simplification, not a precise scientific boundary.

### A second grouping: fast vs. slow

Separately, students were also split into **Fast** and **Slow** groups, based on how quickly (on average, across all puzzles) their eyes reached the correct spot. This is a completely independent measurement from accuracy — a student can be fast *and* wrong, or slow *and* right.

## Discussion Questions

1. Two students both score exactly at the median. Under this method, both get labelled "High." Is that fair? What would change if the rule were "strictly above the median" instead of "at or above"?
2. Not every puzzle is equally hard. If one puzzle is much harder than the rest, how could that distort what "accuracy" really measures? What would you want to check before trusting the accuracy score?
3. A student is labelled "Low-performing" based on a 13-puzzle session that took maybe 15 minutes. What real-world decisions should *not* be made based on a label like that? What would make you more comfortable trusting a performance label?
